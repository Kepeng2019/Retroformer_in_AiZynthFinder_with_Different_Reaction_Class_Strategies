```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# 数据
# -----------------------------
depths = [1, 2, 3, 4, 5]
class_0  = [1.12, 3.98, 19.11, 24.99, 33.00]
class_1  = [1.11, 3.06, 13.19, 19.18, 23.60]
class_3  = [1.16, 3.91, 16.23, 24.76, 27.50]
class_5  = [1.17, 4.15, 20.10, 26.36, 29.17]
class_10 = [1.18, 4.44, 24.50, 31.57, 42.50]

# -----------------------------
# 期刊级绘图设置
# -----------------------------
plt.rcParams.update({
    'font.size': 12,
    'font.family': 'serif',
    'axes.labelsize': 14,
    'axes.titlesize': 16,
    'legend.fontsize': 12,
    'pdf.fonttype': 42,
    'ps.fonttype': 42
})

fig, ax = plt.subplots(figsize=(8, 5.5), dpi=300)

# 配色（ColorBrewer Set2）
colors = ['#66C2A5', '#FC8D62', '#8DA0CB', '#E78AC3', '#A6D854']
markers = ['o', 's', '^', 'D', 'v']

# 绘制折线
lines = [
    (class_0,  'Class 0'),
    (class_1,  'Class 1'),
    (class_3,  'Class 3'),
    (class_5,  'Class 5'),
    (class_10, 'Class 10')
]

for (data, label), color, marker in zip(lines, colors, markers):
    ax.plot(
        depths, data,
        label=label,
        color=color,
        marker=marker,
        linewidth=2,
        markersize=8,
        markeredgecolor='black',
        markeredgewidth=0.6
    )

    # 数据点标注
    for x, y in zip(depths, data):
        ax.annotate(
            f'{y:.2f}',
            xy=(x, y),
            xytext=(0, 6),
            textcoords='offset points',
            ha='center',
            va='bottom',
            fontsize=9
        )

# -----------------------------
# 坐标轴与样式
# -----------------------------
ax.set_xlabel('Search Depth', fontsize=14)
ax.set_ylabel('Routes diversity', fontsize=14)
# ax.set_title('Routes diversity growth across search depths with PaRoutes_50k data', fontsize=16)
ax.set_xticks(depths)
ax.set_xticklabels([f'Depth {d}' for d in depths])
ax.legend(frameon=True, edgecolor='gray')

# 去除上、右边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 水平网格线
ax.yaxis.grid(True, linestyle='--', linewidth=0.5, alpha=0.6)
ax.set_axisbelow(True)

plt.tight_layout()

# -----------------------------
# 保存（PNG + PDF）
# -----------------------------
# plt.savefig('paroutes_depth_lineplot.png', dpi=300, bbox_inches='tight')
# plt.savefig('paroutes_depth_lineplot.pdf', bbox_inches='tight')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# 数据
# -----------------------------
classes = ['Class 0', 'Class 1', 'Class 3', 'Class 5', 'Class 10']
eta = [0.2730, 0.3057, 0.3280, 0.3028, 0.2167]

x = np.arange(len(classes))

# -----------------------------
# 期刊级绘图设置
# -----------------------------
plt.rcParams.update({
    'font.size': 12,
    'font.family': 'serif',
    'axes.labelsize': 14,
    'axes.titlesize': 16,
    'legend.fontsize': 12,
    'pdf.fonttype': 42,
    'ps.fonttype': 42
})

fig, ax = plt.subplots(figsize=(8, 5.5), dpi=300)

# 配色
color = '#4E79A7'

# 折线图
ax.plot(
    x, eta,
    color=color,
    marker='o',
    linewidth=2,
    markersize=10,
    markeredgecolor='black',
    markeredgewidth=0.6,
    markerfacecolor=color
)

# 数据点标注
for xi, yi in zip(x, eta):
    ax.annotate(
        f'{yi:.4f}',
        xy=(xi, yi),
        xytext=(0, 8),
        textcoords='offset points',
        ha='center',
        va='bottom',
        fontsize=11
    )

# -----------------------------
# 坐标轴与样式
# -----------------------------
ax.set_xlabel('Strategy', fontsize=14)
ax.set_ylabel('η(5) ', fontsize=14)
# ax.set_title('PaRoutes_50k — Strategy vs η', fontsize=16)
ax.set_xticks(x)
ax.set_xticklabels(classes)

# Y 轴范围，留出标注空间
ax.set_ylim(0, max(eta) * 1.25)

# 去除上、右边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 水平网格线
ax.yaxis.grid(True, linestyle='--', linewidth=0.5, alpha=0.6)
ax.set_axisbelow(True)

plt.tight_layout()

# -----------------------------
# 保存（PNG + PDF）
# -----------------------------
# plt.savefig('paroutes_50k_eta_lineplot.png', dpi=300, bbox_inches='tight')
# plt.savefig('paroutes_50k_eta_lineplot.pdf', bbox_inches='tight')

plt.show()
```
